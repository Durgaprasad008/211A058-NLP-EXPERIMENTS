```python
# Name: Durga Prasad
# UIN: 211A058
# Experiment 1: Sentiment Analysis Mini Project

import re

positive_words = {
    "good", "great", "excellent", "amazing",
    "happy", "love", "nice", "awesome"
}

negative_words = {
    "bad", "poor", "terrible", "hate",
    "worst", "disappointed", "slow", "awful"
}

print("Name: Durga Prasad | UIN: 211A058")
print("Sentiment word lists are ready.")
```

```python
# Name: Durga Prasad
# UIN: 211A058

def analyze_sentiment(text):
    # Convert to lowercase and extract words without punctuation.
    tokens = re.findall(r"\b[a-z]+\b", text.lower())

    positive_count = 0
    negative_count = 0

    for word in tokens:
        if word in positive_words:
            positive_count += 1
        elif word in negative_words:
            negative_count += 1

    score = positive_count - negative_count

    if score > 0:
        sentiment = "Positive"
    elif score < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    return tokens, positive_count, negative_count, score, sentiment

print("Name: Durga Prasad | UIN: 211A058")
print("Sentiment analysis function is ready.")
```

```python
# Name: Durga Prasad
# UIN: 211A058

reviews = [
    "The product is good and amazing!",
    "The service was bad and slow.",
    "The product arrived today.",
    "The design is good but the quality is poor."
]

print("Name: Durga Prasad | UIN: 211A058")

for review in reviews:
    tokens, pos, neg, score, sentiment = analyze_sentiment(review)

    print("\nReview:", review)
    print("Tokens:", tokens)
    print("Positive words:", pos)
    print("Negative words:", neg)
    print("Score:", score)
    print("Sentiment:", sentiment)
```

```python
# Name: Durga Prasad
# UIN: 211A058

print("Name: Durga Prasad | UIN: 211A058")

review = input("Enter a review: ")

tokens, pos, neg, score, sentiment = analyze_sentiment(review)

print("\nTokens:", tokens)
print("Positive words:", pos)
print("Negative words:", neg)
print("Sentiment score:", score)
print("Final sentiment:", sentiment)
```
